```python
import numpy as np
import matplotlib.pyplot as plt

# Load summaries
von_neumann_data = np.load("data_observable/von_neumann_entropy_summary.npy", allow_pickle=True)
mutual_info_data = np.load("data_observable/mutual_information_summary.npy", allow_pickle=True)

temperature_list = np.arange(0.05, 2.05, 0.05)
delta_t_list = np.arange(0.5, 2.05, 0.05)

S_vN = von_neumann_data[:, 2].astype(float).reshape(len(temperature_list), len(delta_t_list))
MI = mutual_info_data[:, 2].astype(float).reshape(len(temperature_list), len(delta_t_list))

save_dir = r"C:\Users\melor\OneDrive\Desktop\LM-interaction\LM-interaction\plots"

# Von Neumann Entropy
fig, ax = plt.subplots(figsize=(7, 5))
im = ax.imshow(S_vN, aspect='auto', origin='lower',
    extent=[delta_t_list[0], delta_t_list[-1], temperature_list[0], temperature_list[-1]],
    cmap='viridis')
ax.set_xlabel(r'Photon time separation $\Delta t$', fontsize=12)
ax.set_ylabel(r'Spin temperature $T$', fontsize=12)
ax.set_title(r'Von Neumann Entropy $S_{\mathrm{vN}}$', fontsize=13)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.savefig(save_dir + r"\fig_von_neumann_entropy.png", dpi=150, bbox_inches='tight')
plt.show()

# Mutual Information
fig, ax = plt.subplots(figsize=(7, 5))
im = ax.imshow(MI, aspect='auto', origin='lower',
    extent=[delta_t_list[0], delta_t_list[-1], temperature_list[0], temperature_list[-1]],
    cmap='viridis')
ax.set_xlabel(r'Photon time separation $\Delta t$', fontsize=12)
ax.set_ylabel(r'Spin temperature $T$', fontsize=12)
ax.set_title(r'Mutual Information $I(1:2)$', fontsize=13)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.savefig(save_dir + r"\fig_mutual_information.png", dpi=150, bbox_inches='tight')
plt.show()
```
