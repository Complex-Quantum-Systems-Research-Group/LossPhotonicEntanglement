```python
import numpy as np
import matplotlib.pyplot as plt

temperature_list = np.arange(0.05, 2.05, 0.05)
delta_t_list = np.arange(0.5, 2.05, 0.05)

von_neumann_data = np.load("data_observable/von_neumann_entropy_summary.npy", allow_pickle=True)
mutual_info_data = np.load("data_observable/mutual_information_summary.npy", allow_pickle=True)

vn_vals = von_neumann_data[:, 2].astype(float)
mi_vals = mutual_info_data[:, 2].astype(float)

VN_grid = vn_vals.reshape(len(temperature_list), len(delta_t_list))
MI_grid = mi_vals.reshape(len(temperature_list), len(delta_t_list))

extent = [delta_t_list.min(), delta_t_list.max(), temperature_list.min(), temperature_list.max()]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
im1 = axes[0].imshow(VN_grid, origin='lower', aspect='auto', extent=extent, cmap='viridis')
axes[0].set_xlabel('delta_t')
axes[0].set_ylabel('Temperature T')
axes[0].set_title('Von Neumann Entropy')
plt.colorbar(im1, ax=axes[0], label='S')
im2 = axes[1].imshow(MI_grid, origin='lower', aspect='auto', extent=extent, cmap='plasma')
axes[1].set_xlabel('delta_t')
axes[1].set_ylabel('Temperature T')
axes[1].set_title('Mutual Information')
plt.colorbar(im2, ax=axes[1], label='I')
plt.tight_layout()
plt.show()
```
