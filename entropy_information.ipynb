```python
# %% [markdown]
# # Plotting von Neumann Entropy and Mutual Information

# %%
import numpy as np
import matplotlib.pyplot as plt

# Load summary arrays
von_neumann_list = np.load("data_observable/von_neumann_entropy_summary.npy", allow_pickle=True)
mutual_information_list = np.load("data_observable/mutual_information_summary.npy", allow_pickle=True)

# Extract axes
temperature_list = np.unique([x[0] for x in von_neumann_list])
delta_t_list = np.unique([x[1] for x in von_neumann_list])

# Reshape data into 2D grids for plotting
vn_grid = np.array([x[2] for x in von_neumann_list]).reshape(len(temperature_list), len(delta_t_list))
mi_grid = np.array([x[2] for x in mutual_information_list]).reshape(len(temperature_list), len(delta_t_list))

# %%
# Plot von Neumann Entropy
plt.figure(figsize=(8,6))
plt.imshow(vn_grid, origin='lower', aspect='auto',
           extent=[delta_t_list[0], delta_t_list[-1], temperature_list[0], temperature_list[-1]],
           cmap='viridis')
plt.colorbar(label='Von Neumann Entropy')
plt.xlabel('Photon time separation Δt')
plt.ylabel('Spin temperature T')
plt.title('Von Neumann Entropy of Photon Subsystem')
plt.show()

# %%
# Plot Mutual Information
plt.figure(figsize=(8,6))
plt.imshow(mi_grid, origin='lower', aspect='auto',
           extent=[delta_t_list[0], delta_t_list[-1], temperature_list[0], temperature_list[-1]],
           cmap='plasma')
plt.colorbar(label='Mutual Information')
plt.xlabel('Photon time separation Δt')
plt.ylabel('Spin temperature T')
plt.title('Mutual Information Between Photon Modes')
plt.show()
```
